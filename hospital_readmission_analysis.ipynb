# Hospital Readmission Risk Analysis

This project analyzes hospital patient data to identify factors associated with 30-day readmissions.

Goals of the project:
• Explore patterns in patient readmissions
• Identify high-risk diagnosis groups
• Evaluate a predictive model for readmission risk

Dataset: UCI Diabetes 130-Hospital Dataset
Model: Logistic Regression

## Model Performance

Logistic Regression Model Results

Accuracy: 0.68  
ROC-AUC: 0.64  

Hospital readmission prediction is a complex problem and moderate predictive performance is expected using baseline models.

## Key Insights

• Older patients show higher readmission rates.
• Certain diagnosis groups have significantly higher readmission risk.
• The logistic regression model achieved moderate predictive performance (ROC-AUC 0.64).
